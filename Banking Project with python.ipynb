```python
class BankAccount:
    def __init__(self, name, initial_deposit):
        self.opening_cost = 250.0
        self.name = name
        # 250 fee is added to balance but locked from withdrawal
        self.balance = float(initial_deposit) + self.opening_cost

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Successfully deposited {amount} Taka.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        available_to_withdraw = self.balance - self.opening_cost
        if amount > available_to_withdraw:
            print(f"Transaction Denied! You must maintain the {self.opening_cost} Taka opening fee.")
            print(f"Maximum withdrawable: {available_to_withdraw} Taka.")
        elif amount <= 0:
            print("Amount must be positive.")
        else:
            self.balance -= amount
            print(f"Successfully withdrew {amount} Taka.")

    def show_status(self):
        print(f"\n--- Account Details ---")
        print(f"Account Holder: {self.name}")
        print(f"Total Balance: {self.balance} Taka")
        print(f"Locked Fee: {self.opening_cost} Taka")
        print(f"Available to Withdraw: {self.balance - self.opening_cost} Taka")

def main():
    print("Welcome to the Banking Software")
    name = input("Enter your name: ")

    while True:
        try:
            initial = float(input("Enter initial deposit: "))
            
            # Check if deposit is more than 250
            if initial <= 250:
                print("Account opening failed! You must deposit more than 250 Taka for opening your account.")
                continue 
            
            # Success logic
            user = BankAccount(name, initial)
            print("Account created and money added into the account.")
            break 
            
        except ValueError:
            print("Invalid input! Please enter a numeric amount.")

    while True:
        print("\nIf you want to:")
        print("1. Deposit Money")
        print("2. Withdraw Money")
        print("3. Check Balance")
        print("4. Exit")
        
        choice = input("\nSelect an option (1-4): ")

        if choice == '1':
            while True:
                try:
                    amt = float(input("Enter amount to deposit: "))
                except ValueError:
                    print("Invalid input! Please enter a numeric value.")
                else:
                    user.deposit(amt)
                    break 

        elif choice == '2':
            while True:
                try:
                    amt = float(input("Enter amount to withdraw: "))
                except ValueError:
                    print("Invalid input! Please enter a numeric value.")
                else:
                    user.withdraw(amt)
                    break

        elif choice == '3':
            user.show_status()
            
        elif choice == '4':
            print(f"Goodbye, {user.name}!")
            break
        else:
            print("Invalid choice, please try again.")

if __name__ == "__main__":
    main()
```
